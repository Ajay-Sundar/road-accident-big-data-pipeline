# 1. Producing the data  
In this task, we will implement Apache Kafka producers to simulate real-time data streaming. Spark and parallel data processing should not be used in this section, as we are simulating sensors that often lack processing capabilities.  

1. Every 1 second, load 50-100 accident records from the CSV file. You should keep a pointer in the file reading process and advance it per read. The data should be read in chronological order; just the number of records you read every second is a random number between 50 and 100 (inclusive).
2. Add the current timestamp (accident_ts) to the records. 
3. Send your batch of accident data to a Kafka topic with an appropriate name.

```python
# import libraries
from time import sleep
from json import dumps
from kafka import KafkaProducer
import random
import datetime as dt
import csv

# configuration
hostip = "kafka"
topic = "accident_stream"
```

```python
# Read the collision CSV file and store each row as dictionary
def readCSVFile(fileName):
    collision_records = []

    with open(fileName, "rt", encoding="utf-8-sig") as f:
        reader = csv.DictReader(f)

        for row in reader:
            collision_records.append(row)

    return collision_records
```

```python
# Public accident records to Kafka batch by batch
def publish_message(producer_instance, topic_name, data):
    try:
        producer_instance.send(topic_name, data)
        producer_instance.flush()
        print("Message published successfully. Number of records: " + str(len(data)))
        return True

    except Exception as ex:
        print("Exception in publishing message.")
        print(str(ex))
        return False
```

```python
# Create and return a Kafka producer
def connect_kafka_producer():
    _producer = None
    try:
        _producer = KafkaProducer(
            bootstrap_servers=[f"{hostip}:9092"],
            value_serializer=lambda x: dumps(x).encode("ascii"),
            api_version=(3, 9)
        )
    except Exception as ex:
        print("Exception while connecting Kafka.")
        print(str(ex))
    finally:
        return _producer
```

```python
if __name__ == "__main__":
    print("Reading collision records from CSV file...")
    collision_records = readCSVFile("A2B_dataset/A2B/streaming_collision.csv")

    print("Publishing records..")
    producer = connect_kafka_producer()

    rows_to_send = 0
    start_index = 0
    
    try:
        while True:
            # Randomly selected records between 50 and 100 every second
            rows_to_send = random.randint(50, 100)

            # Using the pointer, select the next batch
            end_index = start_index + rows_to_send
            selected_rows = collision_records[start_index:end_index]

            # If no more rows are left, print the statement and stop publishing
            if len(selected_rows) == 0:
                print("All records have been published.")
                break

            # Add current timestamp to every record in the batch
            accident_ts = int(dt.datetime.now().timestamp())

            data = []

            for row in selected_rows:
                record = dict(row)
                record["accident_ts"] = accident_ts # Add timestamp data
                data.append(record)

            success = publish_message(producer, topic, data)

            if success:
                # Move the pointer forward only after successful publishing
                start_index = end_index
                print("Current pointer position:", start_index)
            else:
                # Display kafka failure message
                print("Publishing failed. Pointer cannot be moved.")
                print("Check Kafka connection/topic before continuing.")
                break

            sleep(1)
            
    except KeyboardInterrupt:
        print("Publishing stopped by user.")
```

The collision records are properly published in batches by the Kafka producer to the topic "accident_stream." The number of records that are published at one time is a random number ranging from 50 to 100, as required in the task. The variable "start_index" is the file pointer, which is incremented only after each successful publishing of records to prevent losing any record.

From the output below, it is clear that messages are successfully published, and the file pointer is incremented. The producer could be halted manually once enough records are streamed for testing purpose.
